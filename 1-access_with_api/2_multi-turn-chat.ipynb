```python
## Load the .env file to access environment variables
from dotenv import load_dotenv

load_dotenv()
```

```python
## Import the Anthropic client and initialize it
from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-4-6"
```

```python
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)


def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)


def chat(messages):
    message = client.messages.create(
        model=model,
        max_tokens=1000,
        messages=messages
    )
    return message.content[0].text
```

```python
## Make a starting list of messages
messages = []

## Add in the initial user message "Define quantum computing in 1 sentence"
add_user_message(messages, "Define quantum computing in 1 sentence")
print("1st users messages:", messages)

## Pass the list of messages to chat to get the response
answer = chat(messages)
print("answer:", answer)

## Take the response and add it to the list of messages as an assistant message
add_assistant_message(messages, answer)
print("1st assistant messages:", messages)


## Add the follow-up user message "Write 1 more sentence"
add_user_message(messages, "Write 1 more sentence")
print("2nd users messages:", messages)

## Call chat again to get the response to the follow-up question
final_answer = chat(messages)
print("final answer:", final_answer)
```
